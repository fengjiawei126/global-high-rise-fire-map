本模块只使用 Python 生成 SVG、PDF、TIFF 与 PNG，并同步输出每个图板的 CSV 源数据。

```python
from pathlib import Path # 导入跨平台路径工具
import sys # 导入解释器路径模块
import pandas as pd # 导入表格处理模块
ROOT=Path.cwd().resolve() # 读取当前工作目录
ROOT=ROOT.parent if ROOT.name=='notebooks' else ROOT # 从notebooks目录回到项目根目录
sys.path.insert(0,str(ROOT)) # 将项目根目录加入模块搜索路径
from src.figures import make_figure_1,make_figure_2,make_figure_3,make_figure_4,make_supplementary_figure_1,make_supplementary_figure_2,make_supplementary_figure_3,make_supplementary_figure_4,make_supplementary_figure_5 # 导入全球地图、病例交叉、观察过程、人员后果、消防能力与补充图函数
EVENTS=ROOT/'data/processed/events_enriched.csv' # 指定最终增强事件表
events=pd.read_csv(EVENTS,encoding='utf-8-sig',parse_dates=['event_date']) # 读取增强事件数据
figure=make_figure_1(events,ROOT/'outputs/figures/Figure_1') # 生成论文主图与多格式输出
CASE_CONTROL=ROOT/'data/processed/case_control_weather.csv' # 指定病例交叉气象暴露表
PRIMARY=ROOT/'outputs/tables/conditional_heat_model.csv' # 指定条件Logit主模型结果表
SENSITIVITY=ROOT/'outputs/tables/heat_definition_sensitivity.csv' # 指定热浪定义敏感性结果表
figure_2=make_figure_2(pd.read_csv(CASE_CONTROL,encoding='utf-8-sig'),pd.read_csv(PRIMARY,encoding='utf-8-sig'),pd.read_csv(SENSITIVITY,encoding='utf-8-sig'),ROOT/'outputs/figures/Figure_2') if CASE_CONTROL.exists() and PRIMARY.exists() and SENSITIVITY.exists() else ['pending_real_GEE_results'] # 仅在真实GEE与模型结果齐备时生成病例交叉图
figure_s1=make_supplementary_figure_1(events,ROOT/'outputs/figures/Figure_S1') # 生成来源结构与缺失性补充图
POWER_SENSITIVITY=ROOT/'outputs/tables/nasa_power_heat_definition_sensitivity.csv' # 指定独立气象产品热浪敏感性结果表
figure_s2=make_supplementary_figure_2(pd.read_csv(POWER_SENSITIVITY,encoding='utf-8-sig'),ROOT/'outputs/figures/Figure_S2') if POWER_SENSITIVITY.exists() else ['pending_nasa_power_results'] # 在独立产品结果齐备时生成补充森林图
INJURY_MODEL=ROOT/'outputs/tables/injury_reporting_completeness_model.csv' # 指定受伤报告完整性模型结果表
EVACUATION_MODEL=ROOT/'outputs/tables/evacuation_reporting_completeness_model.csv' # 指定疏散报告完整性模型结果表
figure_3=make_figure_3(events,pd.read_csv(INJURY_MODEL,encoding='utf-8-sig'),pd.read_csv(EVACUATION_MODEL,encoding='utf-8-sig'),ROOT/'outputs/figures/Figure_3') # 生成后果记录与观察过程主图
DEATH_CONSEQUENCE_MODEL=ROOT/'outputs/tables/death_consequence_association_model.csv' # 指定死亡人数探索关联模型结果
INJURY_CONSEQUENCE_MODEL=ROOT/'outputs/tables/injury_consequence_association_model.csv' # 指定受伤人数探索关联模型结果
figure_4=make_figure_4(events,pd.read_csv(DEATH_CONSEQUENCE_MODEL,encoding='utf-8-sig'),pd.read_csv(INJURY_CONSEQUENCE_MODEL,encoding='utf-8-sig'),ROOT/'outputs/figures/Figure_4') # 生成人员后果集中度与探索关联主图
POWER_CASE_CONTROL=ROOT/'data/processed/case_control_weather_nasa_power_v10.csv' # 指定独立气象产品病例对照表
figure_s3=make_supplementary_figure_3(pd.read_csv(POWER_CASE_CONTROL,encoding='utf-8-sig'),events,ROOT/'outputs/figures/Figure_S3') if POWER_CASE_CONTROL.exists() else ['pending_nasa_power_results'] # 生成独立气象产品匹配天气诊断图
CAPACITY_SENSITIVITY=ROOT/'outputs/tables/ctif_fire_service_capacity_sensitivity.csv' # 指定CTIF消防服务能力敏感性结果
figure_s4=make_supplementary_figure_4(events,pd.read_csv(CAPACITY_SENSITIVITY,encoding='utf-8-sig'),ROOT/'outputs/figures/Figure_S4') # 生成消防服务能力覆盖与敏感性补充图
POWER_WINDOW_SENSITIVITY=ROOT/'outputs/tables/nasa_power_event_window_sensitivity.csv' # 指定独立气象产品队列与逐洲留一敏感性结果
figure_s5=make_supplementary_figure_5(pd.read_csv(POWER_WINDOW_SENSITIVITY,encoding='utf-8-sig'),ROOT/'outputs/figures/Figure_S5') # 生成来源时期事件排除与逐洲留一补充图
print({'Figure_1':figure,'Figure_2':figure_2,'Figure_3':figure_3,'Figure_4':figure_4,'Figure_S1':figure_s1,'Figure_S2':figure_s2,'Figure_S3':figure_s3,'Figure_S4':figure_s4,'Figure_S5':figure_s5}) # 显示全部图件输出状态
```
